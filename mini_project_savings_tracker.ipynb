```python


# ==========================================
# NEXFORGE11 ACADEMY
# MONTH 1 - LESSON 01
# MINI PROJECT: SAVINGS TRACKER
# ==========================================

# Starting savings
savings = 75

# Savings goal
goal = 100

# Calculate how much more we need
remaining = goal - savings
print("Still needed:", remaining)

# Add a new deposit
deposit = 10

# Calculate updated savings
new_savings = savings + deposit
print("New savings:", new_savings)

# Calculate how much is left after the deposit
new_remaining = goal - new_savings
print("Remaining to goal:", new_remaining)

# ==========================================
# END OF MINI PROJECT
# ==========================================
```
