```python


# ==========================================
# NEXFORGE11 ACADEMY
# MONTH 1 - LESSON 01
# PYTHON ARITHMETIC PRACTICE
# ==========================================

# 1. ADDITION
# The + symbol adds numbers together.

apples = 5
oranges = 3
total = apples + oranges
print("Addition:")
print(total)


# 2. SUBTRACTION
# The - symbol subtracts one number from another.

chocolates = 10
eaten = 4
remaining = chocolates - eaten
print("Subtraction:")
print(remaining)


# 3. MULTIPLICATION
# The * symbol multiplies numbers.

price = 3
quantity = 4
total = price * quantity
print("Multiplication:")
print(total)


# 4. DIVISION
# The / symbol divides numbers.
# Division normally produces a float.
# A float is a number with a decimal point.

sweets = 24
people = 6
each = sweets / people
print("Division:")
print(each)


# ==========================================
# END OF ARITHMETIC PRACTICE
# ==========================================
```
